```python

import pandas as pd

# Read the processbench parquet file
df = pd.read_parquet("processbench_with_roi.parquet")
```

```python
df.head()
```

```python
# Get unique dataset names from the 'id' column by splitting on '-' and taking the first part
datasets = df['id'].str.split('-').str[0].unique()
print("Unique datasets in the data:")
print(datasets)
```

```python
# Get unique generator names
generators = df['generator'].unique()
print("Unique generators in the data:")
print(generators)
```

```python
df.columns
```

```python
# Get unique PRM model columns (those containing 'PRM' in their name)
prm_cols = [col for col in df.columns if 'PRM' in col]
print("Unique PRM models in the data:")
print(prm_cols)
```

```python
df['relative_percent_increase'] = df.apply(lambda x: (len(x['verbose_roi_step']) - len(x['roi_step'])) / len(x['roi_step']), axis=1)
df['absolute_increase'] = df.apply(lambda x: len(x['verbose_roi_step']) - len(x['roi_step']), axis=1)
df['relative_percent_decrease'] = df.apply(lambda x: (len(x['roi_step']) - len(x['consise_roi_step'])) / len(x['roi_step']), axis=1)
df['absolute_decrease'] = df.apply(lambda x: len(x['roi_step']) - len(x['consise_roi_step']), axis=1)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Filter for incorrect final answers only
df_transformed = df[df['label'] == -1]

# Create a figure with larger size
plt.figure(figsize=(15, 8))

# Get the PRM model columns that contain scores
prm_cols = ['Skywork-o1-Open-PRM-Qwen-2.5-7B',
            'Qwen2.5-Math-PRM-7B']

# Create box plots for each PRM score column, grouped by generator
plot_data = []
for col in prm_cols:
    # Extract scores from the lists in each row
    scores = df_transformed.apply(lambda x: x[col+'--change_numbers_steps'][x['roi_step_num']]-x[col][x['roi_step_num']], axis=1)
    # Calculate total length and bin into categories
    # length = df_transformed['absolute_increase']
    length = df_transformed['roi_step_num']
    # length = pd.qcut(length, q=10)
    temp_df = pd.DataFrame({
        'Length': length,
        'Score': scores,
        'PRM Type': col
    })
    plot_data.append(temp_df)

plot_df = pd.concat(plot_data)

# Create the box plot
sns.boxplot(data=plot_df, x='Length', y='Score', hue='PRM Type')

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Add labels and title
plt.xlabel('Relative Percent Decrease in Length')
plt.ylabel('PRM Score Difference (Verbose - Original)')
plt.title('Distribution of PRM Score Changes for Verbose Steps (Correct Answers Only)')

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Create a figure with larger size
plt.figure(figsize=(15, 8))

# Get the PRM model columns that contain scores
prm_cols = ['Skywork-o1-Open-PRM-Qwen-2.5-7B', 
            'Qwen2.5-Math-PRM-7B']

# Create box plots for each PRM score column, grouped by generator
plot_data = []
for col in prm_cols:
    # Extract scores from the lists in each row
    scores = df.apply(lambda x: np.mean(x[col]), axis=1)
    temp_df = pd.DataFrame({
        'Generator': df['generator'],
        'Score': scores,
        'PRM Type': col
    })
    plot_data.append(temp_df)

plot_df = pd.concat(plot_data)

# Create the box plot
sns.boxplot(data=plot_df, x='Generator', y='Score', hue='PRM Type')

# Rotate x-axis labels for better readability
plt.xticks(rotation=45, ha='right')

# Add labels and title
plt.xlabel('Generator Model')
plt.ylabel('PRM Score')
plt.title('Distribution of PRM Scores by Generator Model')

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Show the plot
plt.show()
```
