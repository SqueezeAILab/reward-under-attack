```python
# Set style for better visualization
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet("attack.parquet")
row = df[df["id"]=="gsm8k-325"].iloc[0]

steps = row["steps"]
aug_steps = row["aug_steps"]
skywork_scores = row["Skywork-o1-Open-PRM-Qwen-2.5-7B"]
aug_skywork_scores = row["Skywork/Skywork-o1-Open-PRM-Qwen-2.5-7B--aug_rewards"]

qwen_scores = row["Qwen2.5-Math-PRM-7B"]
aug_qwen_scores = row["Qwen/Qwen2.5-Math-PRM-7B--aug_rewards"]


plt.figure(figsize=(4,6))

# Prepare data
x = np.arange(len(skywork_scores))
width = 0.35  # Increased width for better separation

# Create grouped bar plot for Skywork
plt.subplot(2, 1, 1)
plt.bar(x - width/2, skywork_scores, width, label='Original', alpha=0.8, color='#1f77b4')
plt.bar(x + width/2, aug_skywork_scores, width, label='Modified', alpha=0.8, color='#ff7f0e')

plt.ylabel('Score', fontsize=12)
plt.title('Skywork Scores Comparison', fontsize=14, pad=15)
plt.xticks(x, [f'{i}' for i in range(len(skywork_scores))])
plt.legend(fontsize=10)
plt.grid(True, linestyle='--', alpha=0.3, axis='y')

# Create grouped bar plot for Qwen
plt.subplot(2, 1, 2)
plt.bar(x - width/2, qwen_scores, width, alpha=0.8, color='#1f77b4')
plt.bar(x + width/2, aug_qwen_scores, width, alpha=0.8, color='#ff7f0e')

plt.xlabel('Step Number', fontsize=12)
plt.ylabel('Score', fontsize=12)
plt.title('Qwen Scores Comparison', fontsize=14, pad=15)
plt.xticks(x, [f'{i}' for i in range(len(skywork_scores))])
plt.grid(True, linestyle='--', alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```
