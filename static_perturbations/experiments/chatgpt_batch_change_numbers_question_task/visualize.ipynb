```python
import matplotlib.pyplot as plt
from difflib import SequenceMatcher

def visualize_diff_words_side_by_side(s1, s2):
    """
    Visualize two input strings side by side, aligned word by word, 
    combining contiguous differences into single cells and highlighting only those cells in red.
    """
    # Split input strings into word lists
    words1 = s1.split()
    words2 = s2.split()
    
    # SequenceMatcher to find matching blocks
    matcher = SequenceMatcher(a=words1, b=words2)
    blocks = matcher.get_matching_blocks()
    
    # Build rows: each row is (left_text, right_text, diff_flag)
    rows = []
    prev_i1, prev_i2 = 0, 0
    
    for block in blocks:
        i1, i2, length = block
        
        # Non-matching slice before this matching block
        non1 = words1[prev_i1:i1]
        non2 = words2[prev_i2:i2]
        if non1 or non2:
            rows.append((" ".join(non1), " ".join(non2), True))
        
        # Matching block: add each matched word as its own row (no highlight)
        for offset in range(length):
            w = words1[i1 + offset]
            rows.append((w, w, False))
        
        prev_i1 = i1 + length
        prev_i2 = i2 + length
    
    # Now render with Matplotlib
    num_rows = len(rows)
    fig, ax = plt.subplots(figsize=(8, num_rows * 0.5))
    ax.axis('off')
    
    # Coordinates: two columns at x=0.25 and x=0.75
    x_positions = [0.25, 0.75]
    # y positions: from num_rows down to 1
    y_positions = list(range(num_rows, 0, -1))
    
    for idx, y in enumerate(y_positions):
        left_text, right_text, is_diff = rows[num_rows - y]  # inverse indexing to match y descending
        # If difference, draw red rectangles behind both cells
        if is_diff:
            rect_left = plt.Rectangle((0, y - 0.4), 0.45, 0.8, facecolor='lightcoral', edgecolor='black')
            rect_right = plt.Rectangle((0.55, y - 0.4), 0.45, 0.8, facecolor='lightcoral', edgecolor='black')
            ax.add_patch(rect_left)
            ax.add_patch(rect_right)
        
        # Draw text centered in each cell
        ax.text(x_positions[0], y, left_text, ha='center', va='center', fontsize=12)
        ax.text(x_positions[1], y, right_text, ha='center', va='center', fontsize=12)
    
    # Column headers
    ax.text(x_positions[0], num_rows + 0.6, "String 1", ha='center', va='center', fontsize=14, weight='bold')
    ax.text(x_positions[1], num_rows + 0.6, "String 2", ha='center', va='center', fontsize=14, weight='bold')
    
    ax.set_xlim(0, 1)
    ax.set_ylim(0.5, num_rows + 1)
    plt.tight_layout()
    plt.show()

# Example usage with the user's specific strings:
visualize_diff_words_side_by_side(
    df.iloc[0]["problem"],
    df.iloc[0]["aug_problem"]
)
```

```python
from difflib import SequenceMatcher

# ANSI escape codes
RED = "\033[31m"
GREEN = "\033[32m"
RESET = "\033[0m"

def highlight_inline_diff(s1, s2):
    """
    Return a single line that shows deletions (from s1) in red and insertions (from s2) in green,
    while leaving unchanged words uncolored.
    """
    words1 = s1.split()
    words2 = s2.split()
    
    matcher = SequenceMatcher(a=words1, b=words2)
    result = []
    
    for tag, i1, i2, j1, j2 in matcher.get_opcodes():
        if tag == "equal":
            # Unchanged words
            result.extend(words1[i1:i2])
        elif tag == "delete":
            # Words only in s1 (deletions): color red
            for w in words1[i1:i2]:
                result.append(f"{RED}{w}{RESET}")
        elif tag == "insert":
            # Words only in s2 (insertions): color green
            for w in words2[j1:j2]:
                result.append(f"{GREEN}{w}{RESET}")
        elif tag == "replace":
            # Words replaced: show old in red then new in green
            for w in words1[i1:i2]:
                result.append(f"{RED}{w}{RESET}")
            for w in words2[j1:j2]:
                result.append(f"{GREEN}{w}{RESET}")
    
    return " ".join(result)

# Example strings
s1 = "The quick brown fox jumps over the lazy dog"
s2 = "The quick red the fox leaped over the dog"

combined = highlight_inline_diff(df.iloc[0]["problem"],
    df.iloc[0]["aug_problem"])
print(combined)
```
