````python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt

colors = ['blue', 'green', 'red', 'purple']
df = pd.read_parquet(f"attack.parquet")
for model in (["Qwen2.5-Math-PRM-7B", "Qwen/Qwen2.5-Math-PRM-7B--aug_rewards"], ["Skywork-o1-Open-PRM-Qwen-2.5-7B", "Skywork/Skywork-o1-Open-PRM-Qwen-2.5-7B--aug_rewards"]):
    for step_num in [0, 1, -2, -1]:
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
        
        # for i, dataset in enumerate(["gsm8k", "math", "olympiadbench", "omnimath"]):
        for i in range(1):
        #     df_sample = df[df["split"] == dataset]
            df_sample = df
            df_sample = df_sample[df_sample["final_answer_correct"] == True]
            df_sample = df_sample[df_sample.apply(lambda row: len(row["aug_steps"]) == len(row["steps"]), axis=1)]


            original_rewards = df_sample[model[0]].apply(lambda x: x[step_num]).to_numpy()
            augmented_rewards = df_sample[model[1]].apply(lambda x: x[step_num]).to_numpy()

            # Original plot
            ax1.hist(original_rewards, bins=30, density=True, alpha=0.3,
                    edgecolor=colors[i], color=colors[i])
            kde = stats.gaussian_kde(original_rewards)
            x_range = np.linspace(original_rewards.min(), original_rewards.max(), 200)
            ax1.plot(x_range, kde(x_range), color=colors[i], lw=2, alpha=0.8,
                    label=f'(Mean: {original_rewards.mean():.3f})')

            # Augmented plot  
            ax2.hist(augmented_rewards, bins=30, density=True, alpha=0.3,
                    edgecolor=colors[i], color=colors[i])
            kde = stats.gaussian_kde(augmented_rewards)
            x_range = np.linspace(augmented_rewards.min(), augmented_rewards.max(), 200)
            ax2.plot(x_range, kde(x_range), color=colors[i], lw=2, alpha=0.8,
                    label=f'(Mean: {augmented_rewards.mean():.3f})')

        ax1.set_xlabel(f'Original Reward -- {model[0]} (Step {step_num})')
        ax1.set_ylabel('Density')
        ax1.legend()

        ax2.set_xlabel(f'Augmented Reward -- {model[1]} (Step {step_num})')
        ax2.set_ylabel('Density')
        ax2.legend()

        plt.tight_layout()
        plt.show()
````

````python
import re
def extract_answer(text):
    boxed = re.search(r"oxed{(.*?)}", text)
    if boxed: 
        return boxed.group(1)

    # Fallback: look for the last numeric line or any number
    lines = [l.strip() for l in text.splitlines() if l.strip()]
    for line in reversed(lines):
        if "answer" in line.lower():
            nums = re.findall(r"-?\d+", line)
            if nums:
                return nums[-1]
        m = re.match(r"^[-+]?\d+", line)
        if m:
            return m.group(0)
    nums = re.findall(r"-?\d+", text)
    return nums[-1] if nums else ""
````

````python
import pandas as pd
df_sample = pd.read_parquet(f"attack.parquet")
df_sample["aug_answer"] = df_sample["aug_steps"].apply(lambda x: extract_answer(x[-1]) if len(x) else "")
df_sample["answer"] = df_sample["steps"].apply(lambda x: extract_answer(x[-1]) if len(x) else "")
````

````python
import re
def extract_wrong_assumption(response):
    # Look for text between <wrong_assumption> tags
    pattern = r'<wrong_assumption>(.*?)</wrong_assumption>'
    match = re.search(pattern, response, re.DOTALL)
    if match:
        return match.group(1).strip()
    return ""

df_sample_ = df_sample[(df_sample["aug_answer"]!=df_sample["answer"]) & (df_sample["final_answer_correct"]==False)]

for idx, row in df_sample_.iterrows():
    steps = row["steps"]
    aug_steps = row["aug_steps"]
    scores = row["Skywork-o1-Open-PRM-Qwen-2.5-7B"]
    aug_scores = row["Skywork/Skywork-o1-Open-PRM-Qwen-2.5-7B--aug_rewards"]

    if len(steps)>0 and len(aug_steps)>0 and ("".join(steps[0].split()) != "".join(aug_steps[0].split())) and (abs(aug_scores[0]-scores[0])<0.1) and scores[0]>0.7:
        print(row["id"])
        print()

        print("Final answer correct?", row.final_answer_correct)
        print()

        print("Problem")
        print("=============")
        print(row["problem"])
        print()

        for idx, (step, aug_step, score, aug_score) in enumerate(zip(steps, aug_steps, scores, aug_scores)):
            
            print(f"Step {idx}")
            print("-------")
            print("S:", step)
            print("\t", score)
            print("A:", aug_step)
            print("\t", aug_score)
            print()
            print()

        wrong_assumption = extract_wrong_assumption(row["aug_response"])
        print("\nWrong Assumption:")
        print("================")
        print(wrong_assumption if wrong_assumption else row["aug_response"].split("\n")[:2])

        print("-"*100)
        print()
````
